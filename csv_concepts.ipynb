### Skiprows/nrows Parameter

```python
pd.read_csv('laptop_prices.xls',usecols=['Company','Product','ScreenW'],skiprows=[2,3]).head()
```

### Skip Bad Lines

```python
pd.read_csv('laptop_prices.xls',sep=';',encoding="latin-1",error_bad_lines=False)
```

### dtypes parameter

```python
pd.read_csv('laptop_prices.xls',dtype={'Ram':float}).head()
```

### handling dates- 

#### if any dataset containing date column it is originally in string data type so by using **parse_dates** it converts into date format which becomes easier to fetch- *can check at its info*

```python
pd.read_csv('laptop_prices.xls',parse_dates=['date'])
```

### Converters

```python
def rename(name):
    if name=='Ultrabook':
        return 'Ultra'
    elif name=='Notebook':
        return 'Note'
    else: return name
```

```python
pd.read_csv('laptop_prices.xls',converters={'TypeName':rename})
```

### na_values parameter

#### wherever there is a value '13.3' get replaced by NaN

```python
pd.read_csv('laptop_prices.xls',na_values=['13.3']).head()
```

### Loading a huge dataset in chunks

```python
dfs=pd.read_csv('laptop_prices.xls',chunksize=300)
```
